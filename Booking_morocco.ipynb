# ETAPE 4 - NORMALISER LES NOMS DES SERVICES

```python
# ============================================================
# ETAPE 4 - NORMALISER LES SERVICES
# ============================================================
links = soup.find_all("a", href=True)
import unicodedata


# 1. LISTE OFFICIELLE DES SERVICES

services_cibles = [
    "Hôtels",
    "Riads",
    "Appartements",
    "Villas",
    "Hostels",
    "Maisons d'hôtes",
    "Camps du désert",
    "Resorts",
    "Campings",
    "Vols",
    "Taxi et transfert aéroport",
    "Location de voiture",
    "Excursions",
    "Expériences",
    "Attractions et activités",
    "Restaurants",
    "Spas et hammams",
    "Massages et soins",
    "Surf et sports nautiques",
    "Quad, buggy et 4x4",
    "Randonnée et trekking",
    "Montgolfière",
    "Balades à dos de chameau",
    "Guides touristiques privés",
    "Visites de médinas",
    "Séances photo et shooting",
    "Cours de cuisine marocaine",
    "Ateliers d'artisanat",
    "Dîners et soirées spectacles",
    "Parcs d'attractions et parcs aquatiques",
    "Beach clubs",
    "Billetterie et pass",
    "Artisanat marocain",
    "Souks et créateurs",
    "Mode et accessoires",
    "Épices et produits du terroir",
    "Conciergerie",
    "Gestion d'appartements",
    "Gestion des calendriers et réservations",
    "Location longue durée de villas et appartements",
    "Organisation de mariages et événements",
    "Traiteur",
    "DJ et animation événementielle",
    "Forum voyageurs",
    "Guides touristiques par ville",
    "Annuaire de professionnels",
    "Recherche de médecins, cliniques et pharmacies",
    "Services juridiques",
    "Services financiers",
    "Professionnels de l'immobilier et du BTP",
    "Vente/location immobilière",
    "Immobilier touristique et hébergements en fraction"
]


# 2. FONCTION DE NORMALISATION

def normalize_text(text):

    # remplacer l'apostrophe typographique
    text = text.replace("’", "'")

    # mettre en minuscules
    text = text.lower()

    # enlever les accents
    text = unicodedata.normalize("NFD", text)
    text = "".join(
        char for char in text
        if unicodedata.category(char) != "Mn"
    )

    # supprimer les espaces inutiles
    text = " ".join(text.split())

    return text


# 3. CREER AUTOMATIQUEMENT LES CORRESPONDANCES EXACTES

aliases = {
    normalize_text(service): service
    for service in services_cibles
}


# 4. AJOUTER LES NOMS DIFFERENTS UTILISES SUR LE SITE

aliases.update({

    "hotel": "Hôtels",
    "hotels": "Hôtels",

    "riad": "Riads",

    "apartment": "Appartements",
    "apartments": "Appartements",

    "villa": "Villas",

    "hostel": "Hostels",

    "maison d'hote": "Maisons d'hôtes",
    "maisons d'hotes": "Maisons d'hôtes",
    "guesthouse": "Maisons d'hôtes",
    "guesthouses": "Maisons d'hôtes",

    "desert camp": "Camps du désert",
    "desert camps": "Camps du désert",

    "resort": "Resorts",

    "camping": "Campings",
    "campsites": "Campings",

    "flights": "Vols",

    "taxi aeroport": "Taxi et transfert aéroport",
    "airport taxi": "Taxi et transfert aéroport",
    "airport transfer": "Taxi et transfert aéroport",

    "location voiture": "Location de voiture",
    "car rental": "Location de voiture",

    "experiences": "Expériences",

    "attractions": "Attractions et activités",
    "activities": "Attractions et activités",

    "restaurant": "Restaurants"
})


# 5. NORMALISER LES LIENS TROUVES

services_normalises = []

services_deja_ajoutes = set()

for a_tag in links:

    nom_site = a_tag.get_text(" ", strip=True)
    href = a_tag.get("href")

    nom_nettoye = normalize_text(nom_site)

    if nom_nettoye in aliases:

        nom_officiel = aliases[nom_nettoye]

        # éviter les doublons
        if nom_officiel not in services_deja_ajoutes:

            services_normalises.append({
                "service": nom_officiel,
                "nom_sur_site": nom_site,
                "href": href
            })

            services_deja_ajoutes.add(nom_officiel)


# 6. AFFICHER LE RESULTAT

for service in services_normalises:
    print(service)
```
